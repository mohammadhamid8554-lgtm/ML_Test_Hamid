## **Adding new features**

```python
df["total_score"] =(df["math score"]) + (df["reading score"] )+ (df["writing score"])
```

```python
df["average_socre"] = round(df["total_score"]/3, 2)
display(df.head())
```

```python
full_math_score = df[df["math score"] == 100]["average_socre"].count()
full_reading_score = df[df["reading score"] == 100]["average_socre"].count()
full_writing_score = df[df["writing score"] == 100]["average_socre"].count()

print(f"Number of students with full marks in Math: {full_math_score}")
print(f"Number of students with full marks in Reading: {full_reading_score}")
print(f"Number of students with full marks in Writing: {full_writing_score}")
```

```python
math_less_20 = df[df["math score"] <= 20]["average_socre"].count()
reading_less_20 = df[df["reading score"] <= 20]["average_socre"].count()
writing_less_20 = df[df["writing score"] <= 20]["average_socre"].count()

print(f"Number of students with less than 20 marks in Math: {math_less_20}")
print(f"Number of students with less than 20 marks in Reading: {reading_less_20}")
print(f"Number of students with less than 20 marks in Writing: {writing_less_20}")
```

## **Data Distribution**
